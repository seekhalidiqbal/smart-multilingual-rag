````python
# ==========================================================
# STEP 5 : DOCUMENT LOADING / INITIALIZE RAG
# ==========================================================

import os
import shutil
from pathlib import Path

# ==========================================================
# LANGCHAIN DOCUMENT LOADERS
# ==========================================================

from langchain_community.document_loaders import (
    PyPDFLoader,
    TextLoader,
    CSVLoader,
    UnstructuredPowerPointLoader,
    UnstructuredFileLoader,
)

# ==========================================================
# OPTIONAL LIBRARIES
# ==========================================================

import pandas as pd
from docx import Document


# ==========================================================
# UPLOAD FOLDER
# ==========================================================

UPLOAD_DIR = "uploaded_documents"

os.makedirs(
    UPLOAD_DIR,
    exist_ok=True
)


# ==========================================================
# GLOBAL VARIABLES
# ==========================================================

documents = []
chunks = []
vector_db = None
retriever = None
qa_chain = None
available_files = []
embedding_model = None


# ==========================================================
# LOAD DOCUMENTS
# ==========================================================

def load_documents(files):

    global documents

    documents = []

    # ------------------------------------------------------
    # CHECK FILES
    # ------------------------------------------------------

    if files is None or len(files) == 0:

        print("⚠️ No files received.")

        return documents


    # ------------------------------------------------------
    # CLEAR OLD UPLOADS
    # ------------------------------------------------------

    if os.path.exists(UPLOAD_DIR):

        shutil.rmtree(
            UPLOAD_DIR
        )

    os.makedirs(
        UPLOAD_DIR,
        exist_ok=True
    )


    # ------------------------------------------------------
    # PROCESS EACH FILE
    # ------------------------------------------------------

    for file in files:

        file_path = None
        filename = "Unknown"

        try:

            # ==================================================
            # GRADIO FILE COMPATIBILITY
            # ==================================================

            if isinstance(file, str):

                file_path = file

            else:

                file_path = getattr(
                    file,
                    "name",
                    str(file)
                )


            # ==================================================
            # FILE NAME
            # ==================================================

            filename = os.path.basename(
                file_path
            )

            save_path = os.path.join(
                UPLOAD_DIR,
                filename
            )


            # ==================================================
            # COPY FILE
            # ==================================================

            if os.path.abspath(
                file_path
            ) != os.path.abspath(
                save_path
            ):

                shutil.copy(
                    file_path,
                    save_path
                )


            # ==================================================
            # EXTENSION
            # ==================================================

            ext = Path(
                filename
            ).suffix.lower()


            print()
            print("-" * 60)
            print(
                f"📄 Processing: {filename}"
            )
            print(
                f"📌 Extension : {ext}"
            )


            # ==================================================
            # PDF
            # ==================================================

            if ext == ".pdf":

                loader = PyPDFLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # DOCX
            #
            # IMPORTANT:
            # Use python-docx directly.
            # This avoids UnstructuredFileLoader,
            # spaCy and click dependency errors.
            # ==================================================

            elif ext == ".docx":

                doc = Document(
                    save_path
                )

                text_parts = []


                # --------------------------------------------------
                # PARAGRAPHS
                # --------------------------------------------------

                for paragraph in doc.paragraphs:

                    text = paragraph.text.strip()

                    if text:

                        text_parts.append(
                            text
                        )


                # --------------------------------------------------
                # TABLES
                # --------------------------------------------------

                for table in doc.tables:

                    for row in table.rows:

                        row_text = " | ".join(
                            cell.text.strip()
                            for cell in row.cells
                        )

                        if row_text.strip():

                            text_parts.append(
                                row_text
                            )


                # --------------------------------------------------
                # COMBINE TEXT
                # --------------------------------------------------

                text = "\n".join(
                    text_parts
                )


                # --------------------------------------------------
                # CREATE LANGCHAIN DOCUMENT
                # --------------------------------------------------

                from langchain_core.documents import (
                    Document as LCDocument
                )

                docs = [
                    LCDocument(
                        page_content=text,
                        metadata={
                            "source": filename
                        }
                    )
                ]


            # ==================================================
            # DOC
            #
            # Old .doc format is not supported by python-docx.
            # ==================================================

            elif ext == ".doc":

                print(
                    f"⚠️ Old .doc format requires "
                    f"additional conversion: {filename}"
                )

                continue


            # ==================================================
            # TXT
            # ==================================================

            elif ext == ".txt":

                loader = TextLoader(
                    save_path,
                    encoding="utf-8",
                    autodetect_encoding=True
                )

                docs = loader.load()


            # ==================================================
            # CSV
            # ==================================================

            elif ext == ".csv":

                loader = CSVLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # PPTX / PPT
            # ==================================================

            elif ext in [
                ".pptx",
                ".ppt"
            ]:

                loader = (
                    UnstructuredPowerPointLoader(
                        save_path
                    )
                )

                docs = loader.load()


            # ==================================================
            # XLSX / XLS
            # ==================================================

            elif ext in [
                ".xlsx",
                ".xls"
            ]:

                dataframe = pd.read_excel(
                    save_path
                )

                text = dataframe.to_string(
                    index=False
                )

                from langchain_core.documents import (
                    Document as ExcelDocument
                )

                docs = [
                    ExcelDocument(
                        page_content=text,
                        metadata={
                            "source": filename
                        }
                    )
                ]


            # ==================================================
            # MARKDOWN
            # ==================================================

            elif ext == ".md":

                loader = TextLoader(
                    save_path,
                    encoding="utf-8",
                    autodetect_encoding=True
                )

                docs = loader.load()


            # ==================================================
            # JSON
            # ==================================================

            elif ext == ".json":

                loader = TextLoader(
                    save_path,
                    encoding="utf-8",
                    autodetect_encoding=True
                )

                docs = loader.load()


            # ==================================================
            # HTML
            # ==================================================

            elif ext == ".html":

                loader = UnstructuredFileLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # XML
            # ==================================================

            elif ext == ".xml":

                loader = TextLoader(
                    save_path,
                    encoding="utf-8",
                    autodetect_encoding=True
                )

                docs = loader.load()


            # ==================================================
            # RTF
            # ==================================================

            elif ext == ".rtf":

                loader = UnstructuredFileLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # ODT
            # ==================================================

            elif ext == ".odt":

                loader = UnstructuredFileLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # ODP
            # ==================================================

            elif ext == ".odp":

                loader = UnstructuredFileLoader(
                    save_path
                )

                docs = loader.load()


            # ==================================================
            # UNSUPPORTED
            # ==================================================

            else:

                print(
                    f"⚠️ Unsupported File: {filename}"
                )

                continue


            # ==================================================
            # ADD DOCUMENTS
            # ==================================================

            if docs:

                documents.extend(
                    docs
                )

                print(
                    f"✅ Loaded: {filename}"
                )

                print(
                    f"📄 Documents loaded: {len(docs)}"
                )

            else:

                print(
                    f"⚠️ No content extracted: {filename}"
                )


        # ======================================================
        # FILE ERROR
        # ======================================================

        except Exception as e:

            import traceback

            print()
            print(
                f"❌ Error loading: "
                f"{filename}"
            )

            print(
                f"Error: {repr(e)}"
            )

            traceback.print_exc()


    # ==========================================================
    # FINAL STATUS
    # ==========================================================

    print()
    print("=" * 60)
    print("STEP 5 COMPLETED")
    print("=" * 60)

    print(
        "📄 Total Documents:",
        len(documents)
    )

    print(
        "📁 Upload Folder:",
        UPLOAD_DIR
    )

    print("=" * 60)


    return documents


# ==========================================================
# STEP 5 READY
# ==========================================================

print(
    "✅ Cell 5 loaded successfully."
)

print(
    "✅ Multi-format document loader ready."
)

print(
    "✅ PDF support ready."
)

print(
    "✅ DOCX support ready."
)

print(
    "✅ TXT support ready."
)

print(
    "✅ CSV support ready."
)

print(
    "✅ XLSX / XLS support ready."
)

print(
    "✅ PPTX / PPT support ready."
)

print(
    "✅ MD support ready."
)

print(
    "✅ JSON support ready."
)

print(
    "✅ HTML / XML support ready."
)

print(
    "✅ RTF / ODT / ODP support ready."
)
````

````python

# ==========================================================
# STEP 6 : SMART CHUNKING (GRADIO RAG V5.2)
# ==========================================================

from langchain_text_splitters import RecursiveCharacterTextSplitter

# ----------------------------------------------------------
# Smart Chunking Function
# ----------------------------------------------------------

def create_chunks():

    global documents
    global chunks

    print("Creating Smart Chunks...")

    if len(documents) == 0:
        raise Exception("No documents loaded. Please upload documents first.")

    text_splitter = RecursiveCharacterTextSplitter(

        chunk_size=800,
        chunk_overlap=150,

        separators=[
            "\n\n",
            "\n",
            ". ",
            "? ",
            "! ",
            "; ",
            ", ",
            " ",
            ""
        ],

        keep_separator=True
    )

    chunks = text_splitter.split_documents(documents)

    print("=" * 60)
    print("✅ STEP 6 COMPLETED")
    print("=" * 60)

    print(f"📄 Total Documents : {len(documents)}")
    print(f"📚 Total Chunks    : {len(chunks)}")

    if len(chunks) > 0:

        print("\nSample Chunk Metadata")
        print(chunks[0].metadata)

        print("\nSample Chunk")
        print("-" * 60)
        print(chunks[0].page_content[:700])

    return chunks
````

````python
# ==========================================================
# STEP 7 : MULTILINGUAL EMBEDDINGS + FAISS
#            GRADIO RAG V5.3
# ==========================================================

import time

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS


# ==========================================================
# BUILD EMBEDDINGS + FAISS VECTOR DATABASE
# ==========================================================

def build_vector_database():

    # ------------------------------------------------------
    # START TIMER
    # ------------------------------------------------------

    start = time.time()


    # ------------------------------------------------------
    # GLOBAL VARIABLES
    # ------------------------------------------------------

    global chunks
    global embedding_model
    global vector_db


    # ------------------------------------------------------
    # CHECK CHUNKS
    # ------------------------------------------------------

    if not chunks:

        raise Exception(
            "No chunks found. Please process documents first."
        )


    print("=" * 60)
    print("🚀 STEP 7 : MULTILINGUAL EMBEDDINGS + FAISS")
    print("=" * 60)

    print(
        f"📄 Documents : {len(documents)}"
    )

    print(
        f"📚 Chunks    : {len(chunks)}"
    )


    # ======================================================
    # LOAD MULTILINGUAL EMBEDDING MODEL
    # ======================================================

    print("\nLoading Multilingual Embedding Model...")

    embedding_model = HuggingFaceEmbeddings(

        model_name="intfloat/multilingual-e5-base",

        model_kwargs={
            "device": "cpu"
        },

        encode_kwargs={
            "normalize_embeddings": True
        }

    )

    print("✅ Embedding Model Loaded")


    # ======================================================
    # CREATE FAISS DATABASE
    # ======================================================

    print(
        "\nCreating FAISS Vector Database..."
    )

    vector_db = FAISS.from_documents(

        documents=chunks,

        embedding=embedding_model

    )

    print(
        "✅ FAISS Database Created"
    )


    # ======================================================
    # SAVE FAISS INDEX
    # ======================================================

    INDEX_PATH = "faiss_index"

    vector_db.save_local(
        INDEX_PATH
    )

    print(
        f"✅ FAISS Index Saved: {INDEX_PATH}"
    )


    # ======================================================
    # END TIMER
    # ======================================================

    end = time.time()

    processing_time = round(
        end - start,
        2
    )


    # ======================================================
    # FINAL STEP 7 REPORT
    # ======================================================

    print("=" * 60)
    print("🎉 STEP 7 COMPLETED")
    print("=" * 60)

    print(
        f"Total Documents : {len(documents)}"
    )

    print(
        f"Total Chunks    : {len(chunks)}"
    )

    print(
        f"Total Vectors   : {vector_db.index.ntotal}"
    )

    print(
        f"Index Saved     : {INDEX_PATH}"
    )

    print(
        f"Time Taken      : {processing_time} sec"
    )

    print("=" * 60)


    # ======================================================
    # RETURN VECTOR DATABASE
    # ======================================================

    return vector_db
````

````python
# new step 9
# ==========================================================
# STEP 9 : SMART MULTI-DOCUMENT RAG V4
# ==========================================================
# FEATURES
# ==========================================================
# + TRUE PAGE COUNT
# + CHUNK/PAGE SEPARATION
# + FILE-BY-FILE COMPARISON
# + STRICT FILE FILTERING
# + PAGE-AWARE RETRIEVAL
# + EXACT FACT RETRIEVAL
# + EXACT QUALIFICATION RETRIEVAL
# + MULTILINGUAL SUPPORT
# + DUPLICATE REMOVAL
# + SOURCE VALIDATION
# + SAFE FILE DETECTION
# + GROQ LLM
# + HALLUCINATION STATISTICS
# ==========================================================


import os
import re

from langchain_community.vectorstores import FAISS


# ==========================================================
# GLOBAL INITIALIZATION
# ==========================================================

if "available_files" not in globals():
    available_files = []

if "documents" not in globals():
    documents = []

if "chunks" not in globals():
    chunks = []

if "vector_db" not in globals():
    vector_db = None

if "retriever" not in globals():
    retriever = None

if "qa_chain" not in globals():
    qa_chain = None


# ==========================================================
# NORMALIZE FILE NAME
# ==========================================================

def normalize_file_name(file_name):

    if not file_name:
        return ""

    return os.path.basename(
        str(file_name)
    ).strip().lower()


# ==========================================================
# PREPARE AVAILABLE FILES
# ==========================================================

def prepare_available_files():

    global documents
    global available_files

    file_names = []

    for doc in documents:

        source = doc.metadata.get(
            "source",
            ""
        )

        if source:

            file_name = os.path.basename(
                str(source)
            )

            if file_name:

                file_names.append(
                    file_name
                )

    available_files = sorted(
        list(
            set(file_names)
        )
    )

    print("=" * 60)
    print("AVAILABLE UPLOADED FILES")
    print("=" * 60)

    if not available_files:

        print("No files found.")

    else:

        for file_name in available_files:

            print(
                "•",
                file_name
            )

    print("=" * 60)

    return available_files


# ==========================================================
# GET ORIGINAL DOCUMENT OBJECTS FROM FILE
# ==========================================================

def get_file_documents(file_name):

    global documents

    if not file_name:

        return []

    target = normalize_file_name(
        file_name
    )

    result = []

    for doc in documents:

        source = normalize_file_name(
            doc.metadata.get(
                "source",
                ""
            )
        )

        if source == target:

            result.append(
                doc
            )

    return result


# ==========================================================
# GET CHUNKS FROM SPECIFIC FILE
# ==========================================================

def get_filtered_documents(file_name):

    global chunks

    if not file_name:

        return []

    target = normalize_file_name(
        file_name
    )

    result = []

    for doc in chunks:

        source = normalize_file_name(
            doc.metadata.get(
                "source",
                ""
            )
        )

        if source == target:

            result.append(
                doc
            )

    return result


# ==========================================================
# TRUE PAGE COUNT
# ==========================================================

def get_true_page_count(file_name):

    matching_documents = (
        get_file_documents(
            file_name
        )
    )

    if not matching_documents:

        return 0

    # ------------------------------------------------------
    # PAGE METADATA
    # ------------------------------------------------------

    page_numbers = set()

    for doc in matching_documents:

        page = doc.metadata.get(
            "page"
        )

        if page is None:

            continue

        try:

            page_numbers.add(
                int(page)
            )

        except Exception:

            pass

    # ------------------------------------------------------
    # UNIQUE PAGE COUNT
    # ------------------------------------------------------

    if page_numbers:

        return len(
            page_numbers
        )

    # ------------------------------------------------------
    # FALLBACK
    # ------------------------------------------------------
    # For DOCX/TXT/CSV/etc. where page metadata may not
    # exist, the processed document objects are counted.
    # ------------------------------------------------------

    return len(
        matching_documents
    )


# ==========================================================
# TRUE TOTAL PAGE COUNT
# ==========================================================

def get_total_true_page_count():

    global available_files

    total_pages = 0

    for file_name in available_files:

        total_pages += (
            get_true_page_count(
                file_name
            )
        )

    return total_pages


# ==========================================================
# GET FILE INFORMATION
# ==========================================================

def get_file_information():

    global available_files

    information = []

    for file_name in available_files:

        page_count = (
            get_true_page_count(
                file_name
            )
        )

        chunk_count = len(
            get_filtered_documents(
                file_name
            )
        )

        information.append({

            "file": file_name,

            "pages": page_count,

            "chunks": chunk_count

        })

    return information


# ==========================================================
# DETECT FILE FROM QUESTION
# ==========================================================

def detect_file_from_question(question):

    global available_files

    if not question:

        return None

    question_lower = (
        str(question)
        .lower()
        .strip()
    )

    # ------------------------------------------------------
    # EXACT FULL FILE NAME
    # ------------------------------------------------------

    for file_name in available_files:

        normalized_name = (
            file_name.lower()
        )

        if normalized_name in question_lower:

            return file_name

    # ------------------------------------------------------
    # FILE NAME WITHOUT EXTENSION
    # ------------------------------------------------------

    for file_name in available_files:

        base_name = os.path.splitext(
            file_name
        )[0].lower()

        if len(base_name) < 4:

            continue

        # Word-boundary style matching
        pattern = (
            r"(?<!\w)"
            + re.escape(base_name)
            + r"(?!\w)"
        )

        if re.search(
            pattern,
            question_lower
        ):

            return file_name

    return None


# ==========================================================
# DETECT COMPARISON QUESTION
# ==========================================================

def is_comparison_question(question):

    if not question:

        return False

    q = (
        str(question)
        .lower()
        .strip()
    )

    comparison_keywords = [

        "compare",
        "comparison",

        "compare documents",
        "compare files",

        "compare document",
        "compare file",

        "compare uploaded documents",
        "compare uploaded files",

        "compare all documents",
        "compare all files",

        "all uploaded documents",
        "all uploaded files",

        "difference between",
        "differences between",

        "similarity between",
        "similarities between",

        "similarities and differences",

        "document comparison",
        "file comparison",

        "compare these documents",
        "compare these files"
    ]

    return any(
        keyword in q
        for keyword in comparison_keywords
    )


# ==========================================================
# DETECT EXACT FACT QUESTION
# ==========================================================

def is_exact_fact_question(question):

    if not question:

        return False

    q = (
        str(question)
        .lower()
        .strip()
    )

    exact_keywords = [

        "email",
        "e-mail",
        "gmail",

        "phone",
        "mobile",
        "telephone",
        "contact number",

        "cnic",

        "roll number",
        "roll no",

        "registration number",
        "registration no",

        "address",

        "date of birth",
        "dob",

        "url",
        "website",

        "name",

        "qualification",

        "degree",

        "date"
    ]

    return any(
        keyword in q
        for keyword in exact_keywords
    )


# ==========================================================
# EXACT QUALIFICATION / TERM RETRIEVAL
# ==========================================================

def retrieve_exact_term_documents(
    question,
    max_results=10
):

    global chunks

    if not question or not chunks:

        return []

    q = (
        str(question)
        .lower()
        .strip()
    )

    # ------------------------------------------------------
    # QUALIFICATION TERM GROUPS
    # ------------------------------------------------------

    term_groups = {

        "fsc": [
            "f.sc.",
            "f.sc",
            "f sc",
            "fsc",
            "f.s.c",
            "intermediate"
        ],

        "bed": [
            "b.ed.",
            "b.ed",
            "b ed",
            "b.e.d",
            "bachelor of education"
        ],

        "med": [
            "m.ed.",
            "m.ed",
            "m ed",
            "m.e.d",
            "master of education"
        ],

        "matric": [
            "matric",
            "matriculation",
            "ssc",
            "s.s.c"
        ],

        "bs": [
            "b.s.",
            "b.sc.",
            "bsc",
            "bachelor of science"
        ],

        "ms": [
            "m.s.",
            "m.sc.",
            "msc",
            "master of science"
        ],

        "mphil": [
            "m.phil",
            "m.phil.",
            "mphil",
            "m phil"
        ],

        "phd": [
            "ph.d",
            "ph.d.",
            "phd",
            "ph d",
            "doctorate"
        ]
    }

    matched_groups = []

    # ------------------------------------------------------
    # DETECT GROUPS
    # ------------------------------------------------------

    for group_name, terms in term_groups.items():

        for term in terms:

            if term in q:

                matched_groups.append(
                    group_name
                )

                break

    if not matched_groups:

        return []

    print(
        "Qualification Groups Detected:",
        matched_groups
    )

    # ------------------------------------------------------
    # SEARCH CHUNKS
    # ------------------------------------------------------

    scored_documents = []

    for doc in chunks:

        text = (
            doc.page_content
            if doc.page_content
            else ""
        )

        text_lower = text.lower()

        score = 0

        for group_name in matched_groups:

            for term in term_groups[
                group_name
            ]:

                if term in text_lower:

                    score += 1

        if score > 0:

            scored_documents.append(
                (
                    score,
                    doc
                )
            )

    # ------------------------------------------------------
    # SORT
    # ------------------------------------------------------

    scored_documents.sort(
        key=lambda item: item[0],
        reverse=True
    )

    results = [
        doc
        for score, doc
        in scored_documents[:max_results]
    ]

    print(
        "Exact-Term Retrieved:",
        len(results)
    )

    return results


# ==========================================================
# EXACT FACT RETRIEVAL
# ==========================================================

def retrieve_exact_fact_documents(
    question,
    max_results=12
):

    global chunks

    if not question or not chunks:

        return []

    q = (
        str(question)
        .lower()
        .strip()
    )

    # ------------------------------------------------------
    # FACT KEYWORDS
    # ------------------------------------------------------

    fact_groups = {

        "email": [
            "email",
            "e-mail",
            "gmail"
        ],

        "phone": [
            "phone",
            "mobile",
            "telephone",
            "contact number"
        ],

        "cnic": [
            "cnic",
            "identity card",
            "id card"
        ],

        "roll": [
            "roll number",
            "roll no"
        ],

        "registration": [
            "registration number",
            "registration no"
        ],

        "address": [
            "address"
        ],

        "dob": [
            "date of birth",
            "dob"
        ],

        "url": [
            "url",
            "website"
        ],

        "name": [
            "name"
        ],

        "qualification": [
            "qualification",
            "degree"
        ],

        "date": [
            "date"
        ]
    }

    matched_groups = []

    for group_name, terms in fact_groups.items():

        for term in terms:

            if term in q:

                matched_groups.append(
                    group_name
                )

                break

    if not matched_groups:

        return []

    print(
        "Exact Fact Groups Detected:",
        matched_groups
    )

    # ------------------------------------------------------
    # SCORE CHUNKS
    # ------------------------------------------------------

    scored_documents = []

    for doc in chunks:

        text = (
            doc.page_content
            if doc.page_content
            else ""
        )

        text_lower = text.lower()

        score = 0

        for group_name in matched_groups:

            for term in fact_groups[
                group_name
            ]:

                if term in text_lower:

                    score += 2

        # --------------------------------------------------
        # EXTRA SCORE FOR COMMON EXACT VALUE PATTERNS
        # --------------------------------------------------

        if "email" in matched_groups:

            if re.search(
                r"\b[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}\b",
                text,
                flags=re.IGNORECASE
            ):

                score += 5

        if "phone" in matched_groups:

            if re.search(
                r"\+?\d[\d\s\-()]{7,}\d",
                text
            ):

                score += 5

        if "cnic" in matched_groups:

            if re.search(
                r"\b\d{5}-\d{7}-\d\b",
                text
            ):

                score += 6

        if score > 0:

            scored_documents.append(
                (
                    score,
                    doc
                )
            )

    # ------------------------------------------------------
    # SORT
    # ------------------------------------------------------

    scored_documents.sort(
        key=lambda item: item[0],
        reverse=True
    )

    results = [
        doc
        for score, doc
        in scored_documents[:max_results]
    ]

    print(
        "Exact-Fact Retrieved:",
        len(results)
    )

    return results


# ==========================================================
# REMOVE DUPLICATES
# ==========================================================

def remove_duplicate_documents(
    retrieved_documents
):

    if not retrieved_documents:

        return []

    unique_documents = []

    seen = set()

    for doc in retrieved_documents:

        source = normalize_file_name(
            doc.metadata.get(
                "source",
                ""
            )
        )

        page = doc.metadata.get(
            "page",
            ""
        )

        content = (
            doc.page_content
            if doc.page_content
            else ""
        )

        key = (
            source,
            str(page),
            content[:300]
        )

        if key not in seen:

            seen.add(
                key
            )

            unique_documents.append(
                doc
            )

    return unique_documents


# ==========================================================
# SORT DOCUMENTS BY FILE + PAGE
# ==========================================================

def sort_documents_by_source(
    retrieved_documents
):

    def sort_key(doc):

        source = normalize_file_name(
            doc.metadata.get(
                "source",
                ""
            )
        )

        page = doc.metadata.get(
            "page",
            999999
        )

        try:

            page = int(page)

        except Exception:

            page = 999999

        return (
            source,
            page
        )

    return sorted(
        retrieved_documents,
        key=sort_key
    )


# ==========================================================
# RETRIEVE FROM SPECIFIC FILE
# ==========================================================

def retrieve_from_specific_file(
    question,
    file_name,
    max_results=10
):

    global vector_db

    file_chunks = (
        get_filtered_documents(
            file_name
        )
    )

    print(
        "File:",
        file_name
    )

    print(
        "Available Chunks:",
        len(file_chunks)
    )

    if not file_chunks:

        return []

    if vector_db is None:

        return []

    try:

        embedding_function = (
            vector_db.embedding_function
        )

        filtered_vector_db = (
            FAISS.from_documents(

                documents=file_chunks,

                embedding=embedding_function
            )
        )

        k = min(
            max_results,
            len(file_chunks)
        )

        fetch_k = min(
            max(20, k),
            len(file_chunks)
        )

        # --------------------------------------------------
        # Use similarity when only very few chunks exist
        # --------------------------------------------------

        if len(file_chunks) <= 3:

            retrieved_docs = (
                filtered_vector_db
                .similarity_search(
                    question,
                    k=k
                )
            )

        else:

            retriever_local = (
                filtered_vector_db
                .as_retriever(

                    search_type="mmr",

                    search_kwargs={

                        "k": k,

                        "fetch_k": fetch_k,

                        "lambda_mult": 0.65
                    }
                )
            )

            retrieved_docs = (
                retriever_local.invoke(
                    question
                )
            )

        return (
            remove_duplicate_documents(
                retrieved_docs
            )
        )

    except Exception as e:

        print(
            "Specific File Retrieval Error:",
            type(e).__name__,
            str(e)
        )

        # --------------------------------------------------
        # Fallback to direct embedding search
        # --------------------------------------------------

        try:

            retrieved_docs = (
                vector_db.similarity_search(
                    question,
                    k=min(
                        12,
                        len(file_chunks)
                    ),
                    filter={
                        "source":
                            file_name
                    }
                )
            )

            return (
                remove_duplicate_documents(
                    retrieved_docs
                )
            )

        except Exception:

            return []


# ==========================================================
# RETRIEVE FROM ALL UPLOADED FILES
# ==========================================================

def retrieve_from_all_files(
    question,
    max_results_per_file=8
):

    global available_files

    all_documents = []

    print("=" * 60)
    print("MULTI-FILE COMPARISON RETRIEVAL")
    print("=" * 60)

    for file_name in available_files:

        print(
            "\nSearching:",
            file_name
        )

        retrieved = (
            retrieve_from_specific_file(

                question,

                file_name,

                max_results=max_results_per_file
            )
        )

        print(
            "Retrieved:",
            len(retrieved)
        )

        all_documents.extend(
            retrieved
        )

    all_documents = (
        remove_duplicate_documents(
            all_documents
        )
    )

    all_documents = (
        sort_documents_by_source(
            all_documents
        )
    )

    print(
        "\nTotal unique retrieved chunks:",
        len(all_documents)
    )

    print("=" * 60)

    return all_documents


# ==========================================================
# BUILD SOURCE CONTEXT
# ==========================================================

def build_source_context(
    retrieved_documents
):

    if not retrieved_documents:

        return ""

    context_parts = []

    for index, doc in enumerate(
        retrieved_documents,
        start=1
    ):

        source = os.path.basename(
            str(
                doc.metadata.get(
                    "source",
                    "Unknown"
                )
            )
        )

        page = doc.metadata.get(
            "page"
        )

        display_page = page

        if page is not None:

            try:

                display_page = (
                    int(page) + 1
                )

            except Exception:

                display_page = page

        header = (
            f"[Retrieved Chunk {index}] "
            f"Source: {source}"
        )

        if display_page is not None:

            header += (
                f" | Page: {display_page}"
            )

        content = (
            doc.page_content
            if doc.page_content
            else ""
        )

        context_parts.append(
            header
            + "\n"
            + content
        )

    return "\n\n".join(
        context_parts
    )


# ==========================================================
# BUILD FILE SUMMARY
# ==========================================================

def build_file_summary():

    information = (
        get_file_information()
    )

    lines = []

    for info in information:

        lines.append(
            f"- {info['file']}: "
            f"{info['pages']} page(s), "
            f"{info['chunks']} text chunk(s)"
        )

    return "\n".join(
        lines
    )


# ==========================================================
# GET LLM
# ==========================================================

def get_rag_llm():

    global qa_chain

    # ------------------------------------------------------
    # Try existing QA chain first
    # ------------------------------------------------------

    try:

        llm = (
            qa_chain
            .combine_documents_chain
            .llm_chain
            .llm
        )

        if llm is not None:

            return llm

    except Exception:

        pass

    # ------------------------------------------------------
    # Groq fallback
    # ------------------------------------------------------

    from langchain_groq import ChatGroq

    llm = ChatGroq(

        model=GROQ_MODEL,

        temperature=0,

        max_tokens=1400
    )

    return llm


# ==========================================================
# SMART RAG QUERY
# ==========================================================################################

def ask_rag(question):

    global qa_chain
    global documents
    global chunks
    global available_files
    global vector_db
    global retriever


    # ======================================================
    # RAG READY CHECK
    # ======================================================

    if qa_chain is None:

        raise Exception(
            "RAG Chain is not ready. "
            "Please process documents first."
        )

    if vector_db is None:

        raise Exception(
            "Vector database is not ready."
        )

    # ======================================================
    # PREPARE FILE LIST
    # ======================================================

    if not available_files:

        prepare_available_files()

    # ======================================================
    # EMPTY QUESTION
    # ======================================================

    if not question or not str(
        question
    ).strip():

        return {

            "result":
                "Please enter a question.",

            "source_documents":
                []
        }

    question = str(
        question
    ).strip()

    question_lower = (
        question.lower()
    )

    print("=" * 60)
    print("SMART RAG QUERY")
    print("=" * 60)

    print(
        "Question:",
        question
    )

    # ======================================================
    # FILE DETECTION
    # ======================================================

    matched_file = (
        detect_file_from_question(
            question
        )
    )

    print(
        "Detected File:",
        matched_file
        if matched_file
        else "None"
    )

    # ======================================================
    # FILE COUNT
    # ======================================================

    file_count_keywords = [

        "how many files",
        "number of files",
        "total files",
        "files are uploaded",
        "files uploaded",
        "uploaded files",
        "how many documents",
        "number of documents"
    ]

    if any(
        keyword in question_lower
        for keyword in file_count_keywords
    ):

        total_files = len(
            available_files
        )

        return {

            "result": (
                f"There are {total_files} "
                f"uploaded file(s)."
            ),

            "source_documents":
                []
        }

    # ======================================================
    # CHUNK COUNT
    # ======================================================

    chunk_count_keywords = [

        "how many chunks",
        "number of chunks",
        "total chunks",
        "chunks generated",
        "generated chunks"
    ]

    if any(
        keyword in question_lower
        for keyword in chunk_count_keywords
    ):

        total_chunks = (
            len(chunks)
            if chunks
            else 0
        )

        return {

            "result": (
                f"{total_chunks} "
                f"chunk(s) were generated."
            ),

            "source_documents":
                []
        }

    # ======================================================
    # TRUE TOTAL PAGE COUNT
    # ======================================================

    total_page_keywords = [

        "how many pages",
        "number of pages",
        "total pages",
        "page count",
        "how many processed pages",
        "processed pages"
    ]

    if any(
        keyword in question_lower
        for keyword in total_page_keywords
    ) and not matched_file:

        total_files = len(
            available_files
        )

        total_pages = (
            get_total_true_page_count()
        )

        return {

            "result": (
                f"There are {total_files} "
                f"uploaded file(s) containing "
                f"{total_pages} processed page(s)."
            ),

            "source_documents":
                []
        }

    # ======================================================
    # SPECIFIC FILE PAGE COUNT
    # ======================================================

    if (
        matched_file
        and
        any(
            keyword in question_lower
            for keyword in [

                "pages",
                "page count",
                "how many page",
                "number of page"
            ]
        )
    ):

        page_count = (
            get_true_page_count(
                matched_file
            )
        )

        source_docs = (
            get_file_documents(
                matched_file
            )
        )

        return {

            "result": (
                f"{matched_file} contains "
                f"{page_count} page(s)."
            ),

            "source_documents":
                source_docs[:1]
        }

    # ======================================================
    # QUERY TYPE DETECTION
    # ======================================================

    comparison_question = (
        is_comparison_question(
            question
        )
    )

    exact_fact_question = (
        is_exact_fact_question(
            question
        )
    )

    print(
        "Comparison:",
        comparison_question
    )

    print(
        "Exact Fact:",
        exact_fact_question
    )

    # ======================================================
    # RETRIEVAL
    # ======================================================

    source_documents = []

    # ------------------------------------------------------
    # PRIORITY 1
    # STRICT FILE RETRIEVAL
    # ------------------------------------------------------

    if matched_file:

        print(
            "STRICT FILE RETRIEVAL"
        )

        source_documents = (
            retrieve_from_specific_file(

                question,

                matched_file,

                max_results=12
            )
        )

        # --------------------------------------------------
        # Add exact retrieval if needed
        # --------------------------------------------------

        if exact_fact_question:

            exact_docs = (
                retrieve_exact_fact_documents(
                    question,
                    max_results=8
                )
            )

            exact_docs = [

                doc
                for doc in exact_docs

                if normalize_file_name(
                    doc.metadata.get(
                        "source",
                        ""
                    )
                )
                ==
                normalize_file_name(
                    matched_file
                )
            ]

            source_documents.extend(
                exact_docs
            )

    # ------------------------------------------------------
    # PRIORITY 2
    # COMPARISON
    # ------------------------------------------------------

    elif comparison_question:

        print(
            "ALL-FILE COMPARISON RETRIEVAL"
        )

        source_documents = (
            retrieve_from_all_files(
                question,
                max_results_per_file=8
            )
        )

    # ------------------------------------------------------
    # PRIORITY 3
    # EXACT FACT RETRIEVAL
    # ------------------------------------------------------

    elif exact_fact_question:

        print(
            "EXACT FACT RETRIEVAL"
        )

        source_documents = (
            retrieve_exact_fact_documents(
                question,
                max_results=12
            )
        )

        # --------------------------------------------------
        # Also use normal semantic retrieval
        # --------------------------------------------------

        try:

            if retriever is not None:

                semantic_docs = (
                    retriever.invoke(
                        question
                    )
                )

                source_documents.extend(
                    semantic_docs
                )

        except Exception:

            pass

    # ------------------------------------------------------
    # PRIORITY 4
    # QUALIFICATION RETRIEVAL
    # ------------------------------------------------------

    else:

        exact_term_docs = (
            retrieve_exact_term_documents(
                question,
                max_results=8
            )
        )

        if exact_term_docs:

            print(
                "EXACT QUALIFICATION RETRIEVAL"
            )

            source_documents.extend(
                exact_term_docs
            )

        # --------------------------------------------------
        # Normal semantic retrieval
        # --------------------------------------------------

        print(
            "GLOBAL MMR RETRIEVAL"
        )

        try:

            if retriever is not None:

                semantic_docs = (
                    retriever.invoke(
                        question
                    )
                )

                source_documents.extend(
                    semantic_docs
                )

            else:

                semantic_docs = (
                    vector_db.similarity_search(
                        question,
                        k=12
                    )
                )

                source_documents.extend(
                    semantic_docs
                )

        except Exception as e:

            print(
                "Retriever Error:",
                type(e).__name__,
                str(e)
            )

            try:

                semantic_docs = (
                    vector_db.similarity_search(
                        question,
                        k=12
                    )
                )

                source_documents.extend(
                    semantic_docs
                )

            except Exception:

                pass

    # ======================================================
    # REMOVE DUPLICATES
    # ======================================================

    source_documents = (
        remove_duplicate_documents(
            source_documents
        )
    )

    # ======================================================
    # SORT BY FILE AND PAGE
    # ======================================================

    source_documents = (
        sort_documents_by_source(
            source_documents
        )
    )

    # ======================================================
    # LIMIT CONTEXT
    # ======================================================

    source_documents = (
        source_documents[:30]
    )

    print(
        "Final Retrieved Chunks:",
        len(source_documents)
    )

    # ======================================================
    # NO RESULTS
    # ======================================================

    if not source_documents:

        return {

            "result":
                "I couldn't find the answer "
                "in the uploaded documents.",

            "source_documents":
                []
        }

    # ======================================================
    # BUILD CONTEXT
    # ======================================================

    context = (
        build_source_context(
            source_documents
        )
    )

    # ======================================================
    # FILE INFORMATION
    # ======================================================

    file_summary = (
        build_file_summary()
    )

    total_files = len(
        available_files
    )

    total_pages = (
        get_total_true_page_count()
    )

    # ======================================================
    # GET LLM
    # ======================================================

    try:

        llm = get_rag_llm()

        print(
            "Groq/RAG Model Loaded"
        )

    except Exception as e:

        print(
            "LLM Loading Error:",
            type(e).__name__,
            str(e)
        )

        return {

            "result":
                f"LLM Error: "
                f"{type(e).__name__}: "
                f"{str(e)}",

            "source_documents":
                source_documents
        }

    # ======================================================
    # FINAL SMART PROMPT
    # ======================================================

    final_prompt = f"""
You are Smart Multilingual AI RAG Assistant.

Answer ONLY from the uploaded document context.

==================================================
CRITICAL GROUNDING RULE
==================================================

Never use outside knowledge.

Never guess.

Never hallucinate.

Never invent facts.

If the requested information is genuinely absent
from the uploaded document context, reply exactly:

"I couldn't find the answer in the uploaded documents."

==================================================
UPLOADED FILE INFORMATION
==================================================

There are exactly:

{total_files} uploaded file(s)

There are:

{total_pages} processed page/document object(s).

TRUE FILE INFORMATION:

{file_summary}

IMPORTANT:

A TEXT CHUNK IS NOT A PAGE.

Never count chunks as pages.

Never treat a chunk as a separate uploaded file.

Pages/chunks belonging to the same filename belong
to the SAME uploaded file.

==================================================
STRICT FILE RULE
==================================================

If the user asks about a specific filename:

Use ONLY information belonging to that filename.

Do NOT use information from another uploaded file.

Do NOT mix facts between files.

==================================================
COMPARISON QUESTIONS
==================================================

If the user asks to compare uploaded documents/files:

1. Treat every filename as one uploaded file.
2. Group all chunks/pages by filename.
3. State the TRUE page count for every file.
4. Never use chunk count as page count.
5. Explain the purpose/content of each file if present.
6. Explain major similarities.
7. Explain major differences.
8. Preserve exact factual information.
9. Do not invent missing information.
10. Do not call processed pages separate files.

Example:

application fee.docx = 1 page
khalid cv.pdf = 2 pages

Correct:

"There are 2 uploaded files containing 3 processed pages."

Incorrect:

"There are 3 uploaded documents."

==================================================
EXACT FACTS
==================================================

For factual fields such as:

Email
Phone
Mobile
Telephone
CNIC
Roll Number
Registration Number
Name
Address
Date
Date of Birth
URL
Website
Qualification
Degree

copy the exact value from the document.

Do not change:

digits,
letters,
symbols,
spacing,
spelling,
or order

unless normal sentence formatting requires it.

==================================================
SOURCE ACCURACY
==================================================

Always associate information with the correct filename.

Example:

CV information -> khalid cv.pdf

Fee application information -> application fee.docx

Never attribute information from one file to another.

==================================================
PAGE ACCURACY
==================================================

When page metadata is available:

Use the page number associated with the source.

Remember:

chunk != page

Multiple chunks may belong to the SAME page.

Multiple pages belong to the SAME uploaded file.

==================================================
LANGUAGE
==================================================

Answer in the SAME language/script style as the user.

English -> English
Urdu -> Urdu
Roman Urdu -> Roman Urdu
Chinese -> Chinese
Japanese -> Japanese
German -> German
Arabic -> Arabic
Hindi -> Hindi

Do not unnecessarily translate the answer.

==================================================
ANSWER STYLE
==================================================

For normal questions:

Give a direct and concise answer.

For comparison questions use:

1. Uploaded Files
2. File-by-File Summary
3. Similarities
4. Differences
5. Important Exact Values

For exact fact questions:

Give the exact value first and identify its source file
when possible.

Do not add outside information.

==================================================
USER QUESTION
==================================================

{question}

==================================================
DOCUMENT CONTEXT
==================================================

{context}

==================================================
ANSWER
==================================================
"""

    # ======================================================
    # GENERATE ANSWER
    # ======================================================

    try:

        response = llm.invoke(
            final_prompt
        )

        answer = (

            response.content

            if hasattr(
                response,
                "content"
            )

            else str(response)
        )

    except Exception as e:

        print("=" * 60)
        print("LLM ERROR")
        print("=" * 60)

        print(
            "Error Type:",
            type(e).__name__
        )

        print(
            "Error Message:",
            str(e)
        )

        print("=" * 60)

        return {

            "result":
                f"LLM Error: "
                f"{type(e).__name__}: "
                f"{str(e)}",

            "source_documents":
                source_documents
        }

    # ======================================================
    # CLEAN ANSWER
    # ======================================================

    if not answer or not str(
        answer
    ).strip():

        answer = (
            "I couldn't find the answer "
            "in the uploaded documents."
        )

    answer = str(
        answer
    ).strip()

    # ======================================================
    # DEBUG SOURCES
    # ======================================================

    print("=" * 60)
    print("RETRIEVED SOURCE DOCUMENTS")
    print("=" * 60)

    for doc in source_documents:

        source = os.path.basename(
            str(
                doc.metadata.get(
                    "source",
                    "Unknown"
                )
            )
        )

        page = doc.metadata.get(
            "page"
        )

        if page is not None:

            try:

                display_page = (
                    int(page) + 1
                )

            except Exception:

                display_page = page

            print(
                f"• {source} | Page {display_page}"
            )

        else:

            print(
                f"• {source}"
            )

    print("=" * 60)

    # ======================================================
    # STATISTICS
    # ======================================================

    try:

        stats_data[
            "answers_generated"
        ] += 1

    except Exception:

        pass

    # ======================================================
    # HALLUCINATION CHECK
    # ======================================================

    try:

        update_hallucination_stats(
            answer,
            source_documents
        )

    except Exception as e:

        print(
            "Hallucination Stats Warning:",
            str(e)
        )

    # ======================================================
    # RETURN
    # ======================================================

    return {

        "result":
            answer,

        "source_documents":
            source_documents
    }


# ==========================================================
# PREPARE FILES AFTER STEP 9 LOAD
# ==========================================================

prepare_available_files()


# ==========================================================
# STEP 9 COMPLETED
# ==========================================================

print("=" * 60)
print("STEP 9 V4 COMPLETED")
print("=" * 60)

print("True Page Count              : READY")
print("Chunk/Page Separation        : READY")
print("File-by-File Comparison      : READY")
print("Strict File Filtering        : READY")
print("Page-Aware Retrieval         : READY")
print("Exact Fact Retrieval         : READY")
print("Exact Qualification Retrieval: READY")
print("Multilingual RAG             : READY")
print("Duplicate Removal            : READY")
print("Source Validation            : READY")
print("Safe File Detection          : READY")
print("Groq LLM Integration         : READY")
print("Hallucination Statistics     : READY")

print("=" * 60)

print(
    "Available Files:",
    len(available_files)
)

print(
    "Total Chunks:",
    len(chunks)
)

print(
    "True Processed Pages:",
    get_total_true_page_count()
)

print("=" * 60)

````

````python
# ==========================================================
# CELL 10
# HALLUCINATION TRACKING — UPDATED / SAFE VERSION
# ==========================================================

# ----------------------------------------------------------
# GLOBAL STATISTICS
# ----------------------------------------------------------

stats_data = {
    "questions_asked": 0,
    "answers_generated": 0,
    "hallucination_checks": 0,
    "grounded_answers": 0,
    "potential_hallucinations": 0,
}


# ----------------------------------------------------------
# UPDATE HALLUCINATION STATISTICS
# ----------------------------------------------------------

def update_hallucination_stats(
    answer,
    source_documents
):

    global stats_data

    stats_data["hallucination_checks"] += 1

    answer_text = str(
        answer or ""
    ).strip()

    # ------------------------------------------------------
    # NO SOURCE
    # ------------------------------------------------------

    if not source_documents:

        stats_data[
            "potential_hallucinations"
        ] += 1

        return False


    # ------------------------------------------------------
    # RAG FALLBACK
    # ------------------------------------------------------

    fallback_phrases = [
        "I couldn't find the answer",
        "I could not find the answer",
        "answer was not found",
        "no answer was found",
        "not found in the uploaded documents"
    ]

    if any(
        phrase.lower() in answer_text.lower()
        for phrase in fallback_phrases
    ):

        stats_data[
            "grounded_answers"
        ] += 1

        return True


    # ------------------------------------------------------
    # SOURCES EXIST
    # ------------------------------------------------------

    stats_data[
        "grounded_answers"
    ] += 1

    return True


# ----------------------------------------------------------
# GET HALLUCINATION STATISTICS
# ----------------------------------------------------------

def get_hallucination_statistics():

    checks = stats_data.get(
        "hallucination_checks",
        0
    )

    grounded = stats_data.get(
        "grounded_answers",
        0
    )

    hallucinations = stats_data.get(
        "potential_hallucinations",
        0
    )


    score = (
        round(
            (grounded / checks) * 100,
            1
        )
        if checks > 0
        else 0
    )


    return {
        "checks": checks,
        "grounded": grounded,
        "hallucinations": hallucinations,
        "score": score
    }


print("✅ Cell 10 loaded successfully.")
print("Hallucination tracking is ready.")
````

````python
# ==========================================================
# STEP 11 — PART 1
# COMPLETE UPDATED BACKEND
#
# FIXES:
# ✔ Questions Asked = Answers Generated
# ✔ Multilingual response
# ✔ RAG Chat
# ✔ Hallucination statistics
# ✔ Source citations
# ✔ TXT / PDF / DOCX downloads
# ✔ Safe Gradio outputs
# ✔ Existing RAG / retrieval code NOT changed
# ==========================================================


# ==========================================================
# INSTALL REQUIRED PACKAGES
# ==========================================================

!pip install -q gTTS reportlab python-docx


# ==========================================================
# IMPORTS
# ==========================================================

import os
import time
import uuid
import gradio as gr

from reportlab.pdfgen import canvas
from docx import Document


# ==========================================================
# LOGO
# ==========================================================

LOGO_PATH = (
    "https://raw.githubusercontent.com/"
    "seekhalidiqbal/rag-assets/main/"
    "logo%20University%20of%20OKara.png"
)


# ==========================================================
# MODEL
# ==========================================================

GROQ_MODEL = globals().get(
    "GROQ_MODEL",
    "openai/gpt-oss-120b"
)


# ==========================================================
# DOWNLOAD DIRECTORY
# ==========================================================

DOWNLOAD_DIR = "downloads"

os.makedirs(
    DOWNLOAD_DIR,
    exist_ok=True
)


# ==========================================================
# DEFAULT STATISTICS
# ==========================================================

default_stats_output = """
| Metric | Status |
|:------|:------:|
| 📂 Uploaded Files | **0** |
| 📄 Processed Documents | **0** |
| 🧩 Generated Chunks | **0** |
| 🔍 Retrieved Chunks | **0** |
| ❓ Questions Asked | **0** |
| 🤖 Answers Generated | **0** |
| 🛡️ Hallucination Checks | **0** |
| ✅ Grounded Answers | **0** |
| ⚠️ Potential Hallucinations | **0** |
| 🎯 Grounding Score | **0%** |
| ⚡ Response Time | **0.00 sec** |
| 🌐 Response Language | **English** |
| 🤖 Model | **openai/gpt-oss-120b** |
"""


# ==========================================================
# SAFE GLOBAL VALUE
# ==========================================================

def get_global_value(name, default=None):

    try:

        value = globals().get(
            name,
            default
        )

        if value is None:
            return default

        return value

    except Exception:

        return default


# ==========================================================
# SAFE CHAT HISTORY
# ==========================================================

def safe_chat_history(chat_history):

    if chat_history is None:
        return []

    try:
        return list(chat_history)

    except Exception:
        return []


# ==========================================================
# CREATE DOWNLOAD FILES
# ==========================================================

def create_download_files(
    question,
    answer,
    sources_text
):

    title = (
        "Smart Multilingual AI RAG Assistant"
    )

    unique_id = uuid.uuid4().hex[:8]


    # ======================================================
    # TXT
    # ======================================================

    txt_path = os.path.join(
        DOWNLOAD_DIR,
        f"rag_answer_{unique_id}.txt"
    )

    with open(
        txt_path,
        "w",
        encoding="utf-8"
    ) as f:

        f.write(title)

        f.write("\n\nQUESTION\n")
        f.write("=" * 70)
        f.write("\n")

        f.write(
            str(question).strip()
        )

        f.write("\n\nANSWER\n")
        f.write("=" * 70)
        f.write("\n")

        f.write(
            str(answer).strip()
        )

        f.write("\n\nSOURCES\n")
        f.write("=" * 70)
        f.write("\n")

        f.write(
            str(sources_text).strip()
        )


    # ======================================================
    # PDF
    # ======================================================

    pdf_path = os.path.join(
        DOWNLOAD_DIR,
        f"rag_answer_{unique_id}.pdf"
    )

    c = canvas.Canvas(
        pdf_path
    )

    width, height = c._pagesize

    x = 40
    y = height - 50


    def pdf_write(
        text,
        font="Helvetica",
        size=11
    ):

        nonlocal y

        c.setFont(
            font,
            size
        )

        for paragraph in str(text).split("\n"):

            words = paragraph.split()

            if not words:

                y -= 16
                continue

            line = ""

            for word in words:

                test_line = (
                    line
                    + " "
                    + word
                ).strip()

                if len(test_line) > 90:

                    if y < 50:

                        c.showPage()

                        y = height - 50

                        c.setFont(
                            font,
                            size
                        )

                    c.drawString(
                        x,
                        y,
                        line
                    )

                    y -= 16

                    line = word

                else:

                    line = test_line

            if line:

                if y < 50:

                    c.showPage()

                    y = height - 50

                    c.setFont(
                        font,
                        size
                    )

                c.drawString(
                    x,
                    y,
                    line
                )

                y -= 16


    pdf_write(
        title,
        "Helvetica-Bold",
        16
    )

    y -= 20

    pdf_write(
        "QUESTION",
        "Helvetica-Bold",
        12
    )

    y -= 5

    pdf_write(question)

    y -= 15

    pdf_write(
        "ANSWER",
        "Helvetica-Bold",
        12
    )

    y -= 5

    pdf_write(answer)

    y -= 15

    pdf_write(
        "SOURCES",
        "Helvetica-Bold",
        12
    )

    y -= 5

    pdf_write(sources_text)

    c.save()


    # ======================================================
    # DOCX
    # ======================================================

    docx_path = os.path.join(
        DOWNLOAD_DIR,
        f"rag_answer_{unique_id}.docx"
    )

    doc = Document()

    doc.add_heading(
        title,
        level=1
    )

    doc.add_heading(
        "Question",
        level=2
    )

    doc.add_paragraph(
        str(question).strip()
    )

    doc.add_heading(
        "Answer",
        level=2
    )

    for paragraph in str(answer).split("\n"):

        if paragraph.strip():

            doc.add_paragraph(
                paragraph.strip()
            )

    doc.add_heading(
        "Sources",
        level=2
    )

    for source in str(sources_text).split("\n"):

        if source.strip():

            doc.add_paragraph(
                source.strip()
            )

    doc.save(
        docx_path
    )


    return (
        txt_path,
        pdf_path,
        docx_path
    )


# ==========================================================
# MULTILINGUAL TRANSLATION
# ==========================================================

def translate_answer_if_needed(
    answer,
    selected_language
):

    if not answer:
        return answer

    selected_language = str(
        selected_language or "English"
    ).strip()


    if selected_language.lower() == "english":

        print(
            "🌐 Response Language: English"
        )

        return str(answer).strip()


    print("=" * 70)
    print("🌐 TRANSLATION REQUEST")
    print(
        "Target Language:",
        selected_language
    )
    print("=" * 70)


    llm = get_global_value(
        "llm",
        None
    )


    if llm is None:

        qa_chain = get_global_value(
            "qa_chain",
            None
        )

        try:

            if qa_chain is not None:

                llm = (
                    qa_chain
                    .combine_documents_chain
                    .llm_chain
                    .llm
                )

                print(
                    "✅ LLM obtained from qa_chain"
                )

        except Exception as e:

            print(
                "⚠️ qa_chain LLM lookup failed:",
                repr(e)
            )


    if llm is None:

        print(
            "❌ Translation failed: LLM not available."
        )

        return answer


    translation_prompt = f"""

You are a professional multilingual AI research assistant.

Translate the following answer into EXACTLY:

{selected_language}

CRITICAL RULES:

1. The final response MUST be written in {selected_language}.
2. Do NOT answer in English unless the target language is English.
3. Do NOT mix languages unnecessarily.
4. Preserve the exact meaning.
5. Do NOT add outside information.
6. Do NOT remove important information.
7. Keep proper names unchanged where appropriate.
8. Keep dates unchanged.
9. Keep numbers unchanged.
10. Keep IDs unchanged.
11. Keep URLs unchanged.
12. Keep file names unchanged.
13. Preserve headings and bullet points.
14. Return ONLY the translated answer.
15. Do NOT explain the translation.
16. Do NOT mention these instructions.

TARGET LANGUAGE:

{selected_language}

ANSWER TO TRANSLATE:

{str(answer).strip()}

"""


    try:

        response = llm.invoke(
            translation_prompt
        )

        translated = getattr(
            response,
            "content",
            response
        )

        translated = str(
            translated
        ).strip()


        if translated.startswith("```"):

            translated = (
                translated
                .replace("```text", "")
                .replace("```markdown", "")
                .replace("```", "")
                .strip()
            )


        if translated:

            print(
                "✅ Translation successful"
            )

            print(
                "🌐 Response Language:",
                selected_language
            )

            return translated


    except Exception as e:

        print(
            "❌ TRANSLATION ERROR:",
            repr(e)
        )


    return answer


# ==========================================================
# RAG CHAT
# ==========================================================

def rag_chat(
    question,
    selected_language,
    chat_history
):

    chat_history = safe_chat_history(
        chat_history
    )


    # ======================================================
    # GLOBAL REFERENCES
    # ======================================================

    vector_db = get_global_value(
        "vector_db",
        None
    )

    qa_chain = get_global_value(
        "qa_chain",
        None
    )

    documents = get_global_value(
        "documents",
        []
    )

    chunks = get_global_value(
        "chunks",
        []
    )

    available_files = get_global_value(
        "available_files",
        []
    )


    # ======================================================
    # EMPTY QUESTION
    # ======================================================

    if (
        question is None
        or not str(question).strip()
    ):

        return (
            chat_history,
            "",
            "⚠️ Please enter a question.",
            default_stats_output,
            gr.update(visible=False),
            gr.update(visible=False),
            gr.update(visible=False)
        )


    clean_question = str(
        question
    ).strip()


    selected_language = str(
        selected_language or "English"
    ).strip()


    print("=" * 70)

    print(
        "QUESTION:",
        clean_question
    )

    print(
        "SELECTED LANGUAGE:",
        selected_language
    )

    print("=" * 70)


    # ======================================================
    # RAG READY CHECK
    # ======================================================

    if (
        vector_db is None
        and qa_chain is None
    ):

        return (
            chat_history,
            "",
            "❌ Please process documents first.",
            default_stats_output,
            gr.update(visible=False),
            gr.update(visible=False),
            gr.update(visible=False)
        )


    start = time.time()


    # ======================================================
    # ASK RAG
    # ======================================================

    try:

        result = ask_rag(
            clean_question
        )

    except Exception as e:

        print(
            "❌ RAG ERROR:",
            repr(e)
        )

        return (
            chat_history,
            "",
            f"❌ RAG Error: {str(e)}",
            default_stats_output,
            gr.update(visible=False),
            gr.update(visible=False),
            gr.update(visible=False)
        )


    # ======================================================
    # NORMALIZE RESULT
    # ======================================================

    if isinstance(result, dict):

        rag_result = result

    else:

        rag_result = {
            "result": str(result),
            "source_documents": []
        }


    # ======================================================
    # ANSWER
    # ======================================================

    answer = rag_result.get(
        "result",
        ""
    )

    if answer is None:

        answer = ""


    answer = str(
        answer
    ).strip()


    if not answer:

        answer = (
            "I couldn't find the answer "
            "in the uploaded documents."
        )


    # ======================================================
    # SOURCE DOCUMENTS
    # ======================================================

    source_documents = rag_result.get(
        "source_documents",
        []
    )


    if source_documents is None:

        source_documents = []


    try:

        retrieved_count = len(
            source_documents
        )

    except Exception:

        retrieved_count = 0


    # ======================================================
    # TRANSLATE
    # ======================================================

    answer = translate_answer_if_needed(
        answer,
        selected_language
    )


    # ======================================================
    # UPDATE COUNTERS
    #
    # ONE SUCCESSFUL ANSWER = ONE QUESTION
    # ======================================================

    stats_data["questions_asked"] += 1

    stats_data["answers_generated"] = (
        stats_data["questions_asked"]
    )


    # ======================================================
    # HALLUCINATION CHECK
    # ======================================================

    try:

        update_hallucination_stats(
            answer,
            source_documents
        )

    except Exception as e:

        print(
            "Hallucination tracking warning:",
            repr(e)
        )


    # ======================================================
    # SOURCE GROUPS
    # ======================================================

    source_groups = {}


    for doc in source_documents:

        try:

            metadata = getattr(
                doc,
                "metadata",
                {}
            )

            if not metadata:

                metadata = {}


            source_value = metadata.get(
                "source",
                metadata.get(
                    "file_name",
                    "Unknown"
                )
            )


            source = os.path.basename(
                str(source_value)
            )


            if not source:

                source = "Unknown"


            page = metadata.get(
                "page",
                None
            )


            if page is not None:

                try:

                    page = int(page) + 1

                except Exception:

                    pass


            if source not in source_groups:

                source_groups[source] = []


            if (
                page is not None
                and page not in source_groups[source]
            ):

                source_groups[source].append(
                    page
                )


        except Exception as e:

            print(
                "Source processing error:",
                repr(e)
            )


    # ======================================================
    # SORT SOURCES
    # ======================================================

    for source in source_groups:

        try:

            source_groups[source] = sorted(
                source_groups[source]
            )

        except Exception:

            pass


    # ======================================================
    # BUILD SOURCE TEXT
    # ======================================================

    sources = []


    for source, pages in source_groups.items():

        if pages:

            page_text = ", ".join(
                str(page)
                for page in pages
            )

            if len(pages) == 1:

                sources.append(
                    f"• {source} (Page {page_text})"
                )

            else:

                sources.append(
                    f"• {source} (Pages {page_text})"
                )

        else:

            sources.append(
                f"• {source}"
            )


    if sources:

        sources_text = (
            "## 📄 Sources\n\n"
            + "\n".join(sources)
        )

    else:

        sources_text = (
            "## 📄 Sources\n\n"
            "• No source found."
        )


    # ======================================================
    # FINAL ANSWER
    # ======================================================

    final_answer = (
        answer
        + "\n\n"
        + sources_text
    )


    # ======================================================
    # HALLUCINATION STATISTICS
    # ======================================================

    try:

        hallucination_stats = (
            get_hallucination_statistics()
        )

    except Exception:

        hallucination_stats = {
            "checks": 0,
            "grounded": 0,
            "hallucinations": 0,
            "score": 0
        }


    checks = hallucination_stats.get(
        "checks",
        0
    )

    grounded = hallucination_stats.get(
        "grounded",
        0
    )

    hallucinations = hallucination_stats.get(
        "hallucinations",
        0
    )

    score = hallucination_stats.get(
        "score",
        0
    )


    # ======================================================
    # MODEL
    # ======================================================

    model_name = get_global_value(
        "GROQ_MODEL",
        "openai/gpt-oss-120b"
    )


    # ======================================================
    # STATISTICS
    # ======================================================

    stats_output = f"""
| Metric | Status |
|:------|:------:|
| 📂 Uploaded Files | **{len(available_files)}** |
| 📄 Processed Documents | **{len(documents)}** |
| 🧩 Generated Chunks | **{len(chunks)}** |
| 🔍 Retrieved Chunks | **{retrieved_count}** |
| ❓ Questions Asked | **{stats_data["questions_asked"]}** |
| 🤖 Answers Generated | **{stats_data["answers_generated"]}** |
| 🛡️ Hallucination Checks | **{checks}** |
| ✅ Grounded Answers | **{grounded}** |
| ⚠️ Potential Hallucinations | **{hallucinations}** |
| 🎯 Grounding Score | **{score}%** |
| ⚡ Response Time | **{round(time.time() - start, 2)} sec** |
| 🌐 Response Language | **{selected_language}** |
| 🤖 Model | **{model_name}** |
"""


    # ======================================================
    # CREATE DOWNLOAD FILES
    # ======================================================

    try:

        (
            txt_path,
            pdf_path,
            docx_path
        ) = create_download_files(
            clean_question,
            final_answer,
            sources_text
        )

    except Exception as e:

        print(
            "Download creation error:",
            repr(e)
        )

        txt_path = None
        pdf_path = None
        docx_path = None


    # ======================================================
    # CHAT HISTORY
    # ======================================================

    chat_history.append(
        {
            "role": "user",
            "content": clean_question
        }
    )

    chat_history.append(
        {
            "role": "assistant",
            "content": final_answer
        }
    )


    # ======================================================
    # STATUS
    # ======================================================

    response_time = round(
        time.time() - start,
        2
    )

    status_message = (
        f"✅ Response Generated "
        f"in {selected_language} "
        f"• {response_time} sec"
    )


    # ======================================================
    # RETURN
    # ======================================================

    return (

        chat_history,

        "",

        status_message,

        stats_output,

        gr.update(
            value=txt_path,
            visible=txt_path is not None
        ),

        gr.update(
            value=pdf_path,
            visible=pdf_path is not None
        ),

        gr.update(
            value=docx_path,
            visible=docx_path is not None
        )
    )


# ==========================================================
# CLEAR CHAT
# ==========================================================

def clear_chat():

    return (

        [],

        "",

        "🧹 Chat Cleared",

        default_stats_output,

        gr.update(
            value=None,
            visible=False
        ),

        gr.update(
            value=None,
            visible=False
        ),

        gr.update(
            value=None,
            visible=False
        )
    )


# ==========================================================
# PROCESS DOCUMENTS
# ==========================================================

def process_and_update_ui(files):

    start_time = time.time()


    if not files:

        return (
            "⚠️ Please select at least one document.",
            default_stats_output
        )


    try:

        status_message = process_documents(
            files
        )

    except Exception as e:

        print("=" * 70)
        print("❌ DOCUMENT PROCESSING ERROR")
        print("=" * 70)
        print(repr(e))

        return (
            f"❌ Processing Error: {str(e)}",
            default_stats_output
        )


    available_files = get_global_value(
        "available_files",
        []
    )

    documents = get_global_value(
        "documents",
        []
    )

    chunks = get_global_value(
        "chunks",
        []
    )

    model_name = get_global_value(
        "GROQ_MODEL",
        "openai/gpt-oss-120b"
    )


    try:

        uploaded_count = len(available_files)

    except Exception:

        uploaded_count = 0


    try:

        document_count = len(documents)

    except Exception:

        document_count = 0


    try:

        chunk_count = len(chunks)

    except Exception:

        chunk_count = 0


    processing_time = round(
        time.time() - start_time,
        2
    )


    if status_message is None:

        status_message = (
            "✅ Documents processed successfully."
        )

    else:

        status_message = str(
            status_message
        )


    hallucination_checks = stats_data.get(
        "hallucination_checks",
        0
    )

    grounded_answers = stats_data.get(
        "grounded_answers",
        0
    )

    potential_hallucinations = stats_data.get(
        "potential_hallucinations",
        0
    )


    stats_output = f"""
| Metric | Status |
|:------|:------:|
| 📂 Uploaded Files | **{uploaded_count}** |
| 📄 Processed Documents | **{document_count}** |
| 🧩 Generated Chunks | **{chunk_count}** |
| 🔍 Retrieved Chunks | **0** |
| ❓ Questions Asked | **{stats_data["questions_asked"]}** |
| 🤖 Answers Generated | **{stats_data["answers_generated"]}** |
| 🛡️ Hallucination Checks | **{hallucination_checks}** |
| ✅ Grounded Answers | **{grounded_answers}** |
| ⚠️ Potential Hallucinations | **{potential_hallucinations}** |
| ⚡ Processing Time | **{processing_time} sec** |
| 🤖 Model | **{model_name}** |
"""


    return (
        status_message,
        stats_output
    )


# ==========================================================
# END OF PART 1
# ==========================================================

print("✅ STEP 11 PART 1 loaded successfully.")
print("✅ Gradio imported successfully.")
print("✅ Multilingual support ready.")
print("✅ Selected-language response system ready.")
print("✅ RAG chat function ready.")
print("✅ Download functions ready.")
````
